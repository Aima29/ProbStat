## BAGIAN A: PROBABILITAS


### 1. Probabilitas Klasik
**Kasus:** Sistem pemantauan kualitas udara di sebuah kota menggunakan 20 sensor IoT. Sebanyak 5 sensor sedang menjalani kalibrasi sehingga tidak dapat digunakan untuk pemantauan. Petugas memilih satu sensor secara acak untuk diperiksa.
Berapakah probabilitas sensor yang terpilih sedang menjalani kalibrasi?


```python
import random

total_sensor = 20
sensor_kalibrasi = 5

# Perhitungan matematis
prob_teori = sensor_kalibrasi / total_sensor
print(f"Probabilitas teoritis: {prob_teori:.4f}")

# Simulasi pemilihan sensor secara acak
jumlah_percobaan = 10000
hasil = 0

for _ in range(jumlah_percobaan):
    sensor = random.randint(1, total_sensor)

    if sensor <= sensor_kalibrasi:
        hasil += 1

prob_simulasi = hasil / jumlah_percobaan
print(f"Probabilitas simulasi: {prob_simulasi:.4f}")
```

**Interpretasi & Kesimpulan:**
Probabilitas terpilihnya sensor yang sedang menjalani kalibrasi adalah 0,25 atau 25%. Artinya, dari 20 sensor yang tersedia, terdapat peluang sebesar 25% untuk memilih sensor yang sedang dikalibrasi. Hasil simulasi dapat sedikit berbeda dari nilai teoritis karena pemilihan dilakukan secara acak. Semakin banyak percobaan, hasil simulasi umumnya semakin mendekati nilai matematis.

### 2. Probabilitas Gabungan
**Kasus:** Pusat pengelolaan Smart City memantau 100 perangkat IoT. Berdasarkan catatan pemantauan, 20 perangkat mengalami gangguan daya, 15 perangkat mengalami gangguan jaringan, dan 5 perangkat mengalami kedua gangguan tersebut.
Jika satu perangkat dipilih secara acak, berapakah probabilitas perangkat tersebut mengalami gangguan daya atau gangguan jaringan?


```python
total = 100
gangguan_daya = 20
gangguan_jaringan = 15
keduanya = 5

prob_daya = gangguan_daya / total
prob_jaringan = gangguan_jaringan / total
prob_keduanya = keduanya / total

prob_gabungan = (
    prob_daya + prob_jaringan - prob_keduanya
)

print(f"Probabilitas gangguan daya : {prob_daya:.4f}")
print(f"Probabilitas gangguan jaringan: {prob_jaringan:.4f}")
print(f"Probabilitas keduanya      : {prob_keduanya:.4f}")
print(f"Probabilitas gabungan      : {prob_gabungan:.4f}")
print(f"Persentase                 : {prob_gabungan * 100:.2f}%")
```

**Interpretasi & Kesimpulan:**
Probabilitas perangkat mengalami gangguan daya atau jaringan adalah 0,30 atau 30%. Lima perangkat yang mengalami kedua gangguan dikurangi agar tidak dihitung dua kali. Dengan demikian, terdapat kemungkinan sebesar 30% perangkat IoT mengalami setidaknya satu dari kedua jenis gangguan tersebut. Informasi ini dapat digunakan untuk memperkirakan jumlah perangkat yang memerlukan pemeriksaan atau pemeliharaan.


### 3. Probabilitas Bersyarat
**Kasus:** Sebuah sistem pemantauan lingkungan menggunakan 50 modul IoT. Dari seluruh modul tersebut, 20 modul mengalami overheating. Dari 20 modul yang mengalami overheating, sebanyak 12 modul juga mengalami penurunan kinerja.
Jika diketahui sebuah modul mengalami overheating, berapakah probabilitas modul tersebut mengalami penurunan kinerja?


```python
total_modul = 50
modul_overheating = 20
overheating_dan_turun_kinerja = 12

prob_bersyarat = (
    overheating_dan_turun_kinerja / modul_overheating
)

print(
    "Probabilitas penurunan kinerja jika overheating:",
    f"{prob_bersyarat:.4f}"
)
print(f"Persentase: {prob_bersyarat * 100:.2f}%")
```

**Interpretasi & Kesimpulan:**
Jika sebuah modul diketahui mengalami overheating, probabilitas modul tersebut mengalami penurunan kinerja adalah 0,60 atau 60%. Berdasarkan data kasus, 12 dari 20 modul yang mengalami overheating juga mengalami penurunan kinerja. Oleh karena itu, overheating dapat menjadi indikator yang perlu diperhatikan dalam pemantauan perangkat IoT.


### 4. Teorema Bayes
**Kasus:** Sistem keamanan Smart City menggunakan sensor IoT untuk mendeteksi asap dan memberikan peringatan dini kebakaran. Berdasarkan asumsi pengujian, probabilitas terjadinya kebakaran adalah 2%.
Sensor memiliki kemampuan mendeteksi kebakaran sebesar 95%. Namun, ketika tidak terjadi kebakaran, sensor tetap dapat memberikan alarm positif palsu sebesar 5%.
Jika sensor memberikan alarm positif, berapakah probabilitas bahwa benar-benar terjadi kebakaran?


```python
prob_kebakaran = 0.02
prob_tidak_kebakaran = 1 - prob_kebakaran

prob_alarm_jika_kebakaran = 0.95
prob_alarm_tanpa_kebakaran = 0.05

prob_kebakaran_jika_alarm = (
    prob_alarm_jika_kebakaran * prob_kebakaran
) / (
    prob_alarm_jika_kebakaran * prob_kebakaran
    + prob_alarm_tanpa_kebakaran * prob_tidak_kebakaran
)

print(
    f"Probabilitas kebakaran jika alarm positif: "
    f"{prob_kebakaran_jika_alarm:.4f}"
)
print(
    f"Persentase: {prob_kebakaran_jika_alarm * 100:.2f}%"
)
```

### 5. Perbandingan Hasil Matematis dan Python
**Kasus:** Petugas Smart City ingin memastikan apakah hasil perhitungan probabilitas pemilihan sensor kualitas udara sesuai dengan hasil simulasi Python. Dari 20 sensor, 5 sensor sedang dikalibrasi.
Perhitungan matematis memberikan probabilitas sebesar 0,25. Untuk memeriksanya, dilakukan simulasi pemilihan sensor sebanyak 100.000 kali.

```python
import random

total_sensor = 20
sensor_kalibrasi = 5

# Hasil matematis
prob_matematis = sensor_kalibrasi / total_sensor

# Simulasi percobaan acak
jumlah_percobaan = 100000
jumlah_kalibrasi = 0

for _ in range(jumlah_percobaan):
    sensor = random.randint(1, total_sensor)

    if sensor <= sensor_kalibrasi:
        jumlah_kalibrasi += 1

prob_python = jumlah_kalibrasi / jumlah_percobaan

print(f"Hasil matematis : {prob_matematis:.4f}")
print(f"Hasil simulasi  : {prob_python:.4f}")
print(f"Selisih         : {abs(prob_matematis - prob_python):.4f}")
```

**Interpretasi & Kesimpulan:**
Hasil matematis memberikan probabilitas teoritis sebesar 0,25, sedangkan hasil simulasi dapat sedikit berbeda karena proses pemilihan dilakukan secara acak. Perbedaan tersebut merupakan variasi hasil simulasi, bukan berarti rumus matematis salah. Semakin banyak percobaan dilakukan, hasil simulasi umumnya semakin mendekati nilai teoritis.


## BAGIAN B: DISTRIBUSI PROBABILITAS


### 1. Distribusi Binomial
**Kasus:** Sebuah perangkat IoT mengirimkan data ke server sebanyak 10 kali. Probabilitas setiap pengiriman berhasil adalah 0,9. Setiap pengiriman diasumsikan saling independen.
Berapakah probabilitas tepat 8 pengiriman berhasil? Tentukan juga nilai harapan dan varians jumlah pengiriman yang berhasil.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

n = 10
p = 0.9
x_target = 8

# Probabilitas, nilai harapan, dan varians
prob = binom.pmf(x_target, n, p)
mean = binom.mean(n, p)
variance = binom.var(n, p)

print(f"P(X = 8): {prob:.4f}")
print(f"Nilai harapan: {mean:.2f}")
print(f"Varians: {variance:.2f}")

# Visualisasi distribusi Binomial
x = np.arange(0, n + 1)
y = binom.pmf(x, n, p)

plt.figure(figsize=(8, 4))
plt.bar(x, y, color="skyblue")
plt.bar(x_target, prob, color="orange", label="P(X = 8)")
plt.title("Distribusi Binomial Pengiriman Data IoT")
plt.xlabel("Jumlah pengiriman berhasil")
plt.ylabel("Probabilitas")
plt.xticks(x)
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
```

**Interpretasi & Kesimpulan:**
Probabilitas tepat 8 pengiriman berhasil adalah sekitar 19,37%. Nilai harapan sebesar 9 menunjukkan bahwa dari 10 pengiriman, rata-rata 9 pengiriman diperkirakan berhasil. Varians sebesar 0,9 menunjukkan tingkat penyebaran jumlah keberhasilan dari nilai harapannya. Distribusi Binomial sesuai digunakan karena jumlah percobaan tetap dan setiap percobaan memiliki dua kemungkinan hasil, yaitu berhasil atau gagal.

### 2. Distribusi Poisson
**Kasus:** Sebuah sistem Smart City menggunakan sensor lalu lintas untuk menghitung kendaraan yang melewati persimpangan. Rata-rata terdapat 7 kendaraan yang melewati sensor setiap menit.
Berapakah probabilitas tepat 5 kendaraan melewati sensor dalam satu menit? Tentukan nilai harapan dan variansnya.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

lam = 7
x_target = 5

# Probabilitas, nilai harapan, dan varians
prob = poisson.pmf(x_target, lam)
mean = poisson.mean(lam)
variance = poisson.var(lam)

print(f"P(X = 5): {prob:.4f}")
print(f"Nilai harapan: {mean:.2f}")
print(f"Varians: {variance:.2f}")

# Visualisasi distribusi Poisson
x = np.arange(0, 16)
y = poisson.pmf(x, lam)

plt.figure(figsize=(8, 4))
plt.bar(x, y, color="skyblue")
plt.bar(x_target, prob, color="orange", label="P(X = 5)")
plt.title("Distribusi Poisson Kedatangan Kendaraan")
plt.xlabel("Jumlah kendaraan per menit")
plt.ylabel("Probabilitas")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()

# Simulasi
simulasi = poisson.rvs(mu=lam, size=10000, random_state=42)
print(f"Rata-rata simulasi: {np.mean(simulasi):.2f}")
print(f"Varians simulasi: {np.var(simulasi):.2f}")
```

**Interpretasi & Kesimpulan:**
Probabilitas tepat 5 kendaraan melewati sensor dalam satu menit adalah sekitar 12,77%. Nilai harapan dan varians masing-masing sebesar 7. Artinya, rata-rata jumlah kendaraan yang terdeteksi adalah 7 kendaraan per menit, dengan variasi jumlah kedatangan yang juga ditentukan oleh parameter tersebut. Distribusi Poisson dapat digunakan untuk memodelkan jumlah kejadian dalam interval waktu tertentu, seperti kedatangan kendaraan atau permintaan layanan.


### 3. Distribusi Normal
**Kasus:** Sensor IoT pada Smart Farm memantau suhu lingkungan. Suhu diasumsikan mengikuti distribusi Normal dengan rata-rata 24°C dan simpangan baku 1,5°C.
Berapakah probabilitas suhu melebihi 27°C? Tentukan nilai harapan dan variansnya.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

mu = 24
sigma = 1.5
x_target = 27

# Probabilitas, nilai harapan, dan varians
prob = stats.norm.sf(x_target, loc=mu, scale=sigma)
mean = stats.norm.mean(loc=mu, scale=sigma)
variance = stats.norm.var(loc=mu, scale=sigma)

print(f"P(X > 27°C): {prob:.4f}")
print(f"Nilai harapan: {mean:.2f}°C")
print(f"Varians: {variance:.2f}")

# Visualisasi distribusi Normal
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 500)
y = stats.norm.pdf(x, loc=mu, scale=sigma)

plt.figure(figsize=(8, 4))
plt.plot(x, y, color="blue", label="Kurva Normal")
plt.fill_between(
    x, y, where=(x >= x_target),
    color="orange", alpha=0.5,
    label="Area P(X > 27°C)"
)
plt.axvline(mu, color="green", linestyle="--", label=f"Mean = {mu}°C")
plt.axvline(x_target, color="red", linestyle="--", label="Batas = 27°C")
plt.title("Distribusi Normal Suhu Smart Farm")
plt.xlabel("Suhu (°C)")
plt.ylabel("Kepadatan probabilitas")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# Simulasi
simulasi = stats.norm.rvs(
    loc=mu, scale=sigma, size=10000, random_state=42
)
print(f"Rata-rata simulasi: {np.mean(simulasi):.2f}")
print(f"Varians simulasi: {np.var(simulasi):.2f}")
print(f"Proporsi suhu > 27°C: {np.mean(simulasi > x_target):.4f}")
```

**Interpretasi & Kesimpulan:**
Probabilitas suhu melebihi 27°C adalah sekitar 0,0228 atau 2,28%. Nilai harapan sebesar 24°C menunjukkan suhu rata-rata yang diperkirakan, sedangkan varians 2,25 menunjukkan penyebaran suhu terhadap rata-ratanya. Distribusi Normal dapat digunakan untuk menganalisis pengukuran lingkungan yang cenderung terkumpul di sekitar nilai rata-rata.

### 4. Distribusi Eksponensial
**Kasus:** Sebuah perangkat IoT pada sistem Smart City memiliki rata-rata waktu operasi hingga mengalami kegagalan sebesar 4 tahun. Waktu hingga kegagalan diasumsikan mengikuti distribusi Eksponensial.
Berapakah probabilitas perangkat tetap beroperasi lebih dari 4 tahun? Tentukan nilai harapan dan variansnya.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import expon

rata_rata = 4
lambda_rate = 1 / rata_rata
t_target = 4

# Probabilitas, nilai harapan, dan varians
prob = expon.sf(
    t_target, scale=1 / lambda_rate
)
mean = expon.mean(scale=1 / lambda_rate)
variance = expon.var(scale=1 / lambda_rate)

print(f"P(X > 4 tahun): {prob:.4f}")
print(f"Nilai harapan: {mean:.2f} tahun")
print(f"Varians: {variance:.2f}")

# Visualisasi distribusi Eksponensial
x = np.linspace(0, 16, 500)
y = expon.pdf(x, scale=1 / lambda_rate)

plt.figure(figsize=(8, 4))
plt.plot(x, y, color="blue", label="Kurva Eksponensial")
plt.fill_between(
    x, y, where=(x >= t_target),
    color="orange", alpha=0.5,
    label="Area P(X > 4 tahun)"
)
plt.axvline(
    t_target, color="red", linestyle="--",
    label="Batas = 4 tahun"
)
plt.title("Distribusi Eksponensial Waktu Operasi IoT")
plt.xlabel("Waktu operasi (tahun)")
plt.ylabel("Kepadatan probabilitas")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# Simulasi
simulasi = expon.rvs(
    scale=1 / lambda_rate, size=10000, random_state=42
)
print(f"Rata-rata simulasi: {np.mean(simulasi):.2f}")
print(f"Varians simulasi: {np.var(simulasi):.2f}")
print(f"Proporsi waktu > 4 tahun: {np.mean(simulasi > t_target):.4f}")
```

**Interpretasi & Kesimpulan:**
Probabilitas perangkat tetap beroperasi lebih dari 4 tahun adalah sekitar 0,3679 atau 36,79%. Nilai harapan sebesar 4 tahun menunjukkan rata-rata waktu operasi perangkat, sedangkan varians sebesar 16 tahun kuadrat menunjukkan penyebaran waktu operasi. Distribusi Eksponensial dapat digunakan untuk memodelkan waktu tunggu hingga suatu kejadian terjadi, misalnya waktu hingga perangkat IoT mengalami kegagalan.
